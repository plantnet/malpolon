```python
def clean_urls(url):
    """
    Clean URLs by removing server prefix.
    """
    if isinstance(url, str):
        url = '/'.join(url.split('/')[-5:])
    else:
        url = ''
    return url
    
def 

def preprocess_lucas(df):
    """
    Preprocess the Lucas dataset.
    """
    cols_useful = ['id', 'point_id', 'year', 'nuts0', 'nuts1', 'nuts2', 'nuts3', 'gps_proj', 'gps_lat', 'gps_long', 'gps_altitude', 'gps_ew', 'file_path_gisco_north', 'file_path_gisco_south', 'file_path_gisco_east', 'file_path_gisco_west', 'file_path_gisco_point']
    cols_fps = ['file_path_gisco_north', 'file_path_gisco_south', 'file_path_gisco_east', 'file_path_gisco_west', 'file_path_gisco_point']

    df_clean = df[cols_useful]
    for col_fp in cols_fps:
        df_clean.loc[:, col_fp] = df_clean[col_fp].apply(clean_urls)
    df_clean['file_path_gisco_cover'] = df_clean['file_path_gisco_point'].copy()
    df_clean['file_path_gisco_cover'] = df_clean['file_path_gisco_cover'].apply(lambda x: str(Path(x).parent / Path(Path(x).stem[:-1] + 'C' + Path(x).suffix)))

    # # Remove duplicates
    # df = df.drop_duplicates()

    # # Remove rows with NaN values in 'image' or 'label' columns
    # df = df.dropna(subset=['image', 'label'])

    # # Convert 'label' column to string
    # df['label'] = df['label'].astype(str)

    # # Convert 'image' column to string
    # df['image'] = df['image'].astype(str)

    return df_clean
```

```python
df_clean = preprocess_lucas(df)
```
